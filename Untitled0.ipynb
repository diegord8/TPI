```python
from google.colab import files
import pandas as pd

archivos = files.upload()
nombre_csv = next(iter(archivos))
df = pd.read_csv(nombre_csv)

print(df.shape)
display(df.head())
```

```python

from sklearn.model_selection import GroupShuffleSplit
from sklearn.preprocessing import MinMaxScaler
import tensorflow as tf

# Las seis variables del modelo
columnas = [
    'temperatura_ambiente_c',
    'humedad_relativa_pct',
    'corriente_a',
    'lectura_radiacion_simulada_usvh',
    'aceleracion_x_g',
    'aceleracion_y_g'
]

# Separar entrenamiento y prueba
train = df[df['particion_temporal'] == 'entrenamiento']
test = df[df['particion_temporal'] == 'prueba']

# Separar una parte para validación
separador = GroupShuffleSplit(
    n_splits=1, test_size=0.2, random_state=42
)
idx_ent, idx_val = next(separador.split(
    train, groups=train['id_par_escenario']
))

ent = train.iloc[idx_ent]
val = train.iloc[idx_val]

# Normalizar las variables
escalador = MinMaxScaler()

X_ent = escalador.fit_transform(ent[columnas])
X_val = escalador.transform(val[columnas])
X_test = escalador.transform(test[columnas])

y_ent = ent['riesgo_simulado']
y_val = val['riesgo_simulado']
y_test = test['riesgo_simulado']

# Crear red neuronal similar a Shazril
tf.keras.utils.set_random_seed(42)

modelo = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(6,)),
    tf.keras.layers.Dense(6, activation='relu'),
    tf.keras.layers.Dense(6, activation='relu'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])

modelo.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Entrenar
historial = modelo.fit(
    X_ent, y_ent,
    validation_data=(X_val, y_val),
    epochs=40,
    batch_size=32,
    verbose=1
)
```
